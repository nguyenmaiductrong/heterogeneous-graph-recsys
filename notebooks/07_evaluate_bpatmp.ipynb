## 7. Evaluate

```python
!rm -rf /content/checkpoints
```

```python
import wandb

api = wandb.Api()

entity = "nguyenmaiductrong37-h-c-vi-n-c-ng-ngh-b-u-ch-nh-vi-n-th-ng"
project = "bpatmp-recsys"
artifact_name = "bpatmp-final-l1"

path_to_artifact = f"{entity}/{project}/{artifact_name}"
versions = api.artifact_versions("model", path_to_artifact)

for v in versions:
    label = v.aliases[0] if v.aliases else v.version
    v.download(root=f"/content/checkpoints/{label}")
    print(f"Đã tải thành công version: {label}")
```

```python
import subprocess
from pathlib import Path

for ckpt in sorted(Path("/content/checkpoints").rglob("*.pt")):
    print(f"\n>>> {ckpt}")
    result = subprocess.run(
        ["python", "-u", "scripts/evaluate.py", "--split", "test", "--checkpoint", str(ckpt)],
        capture_output=True, text=True
    )
    print(result.stdout)
    if result.stderr:
        print(result.stderr)
```
